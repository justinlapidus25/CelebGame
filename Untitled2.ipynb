```python
import pandas as pd
import re

with open("celebs.txt.rtf", "r", encoding="utf-8", errors="ignore") as f:
    text = f.read()

# Remove basic RTF formatting
text = re.sub(r"{\\.*?}", "", text)
text = re.sub(r"\\[a-z]+\d* ?", "", text)
text = text.replace("{", "").replace("}", "")

names = text.split(",")

cleaned_names = []

for name in names:
    name = name.strip()
    name = re.sub(r"\s+", " ", name)

    if len(name.split()) >= 2:
        cleaned_names.append(name)

df = pd.DataFrame(cleaned_names, columns=["Name"])

df = df.drop_duplicates()

df.to_csv("my_celebs.csv", index=False)

print(df.head())
print(len(df))
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

all_names = []

# 25 pages on the site
for page in range(1, 1000):

    url = f"https://www.listchallenges.com/1000-actors-and-actresses/list/{page}"

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    names = soup.find_all("div", class_="item-name")

    for name in names:
        cleaned = name.get_text(strip=True)

        if cleaned:
            all_names.append(cleaned)

# Create dataframe
df = pd.DataFrame(all_names, columns=["Name"])

# Remove duplicates
df = df.drop_duplicates()

# Save CSV
df.to_csv("actors_1000.csv", index=False)

print(df.head())
print(len(df))
```

```python
import pandas as pd

# Load current file
old_df = pd.read_csv("my_celebs.csv")

# Load new celebrity dataset
celeb_df = pd.read_csv("Celebrity.csv")

# Rename column to Name
celeb_df = celeb_df.rename(columns={"name": "Name"})

# Keep ONLY the Name column
old_df = old_df[["Name"]]
celeb_df = celeb_df[["Name"]]

# Combine datasets
df = pd.concat([old_df, celeb_df], ignore_index=True)

# Clean names
df["Name"] = df["Name"].astype(str).str.strip()

# Remove duplicates
df = df.drop_duplicates(subset=["Name"])

# Remove blank rows
df = df[df["Name"] != ""]

# Save updated dataset
df.to_csv("my_celebs.csv", index=False)

print(df.head())
print(len(df))
```
